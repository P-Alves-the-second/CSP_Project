# Modelos (Models)

Este notebook define as classes principais utilizadas no projeto CSP.  
Nesta parte, é apresentada a classe **`Aula`**, que representa uma aula com as suas informações básicas:

- `sala` — sala onde a aula ocorre  
- `bloco` — número do bloco horário da aula


```python
class Aula:
    def __init__(self, sala, bloco):
        """
        Representa uma aula.

        Parâmetros:
            sala (str): Nome ou identificador da sala.
            bloco (int): Número do bloco horário em que a aula ocorre.
        """
        self.sala = sala
        self.bloco = bloco
```

## Teste rápido da classe `Aula`
Cria-se um objeto `Aula` para verificar se a classe está a funcionar corretamente.


```python
aula_exemplo = Aula("SalaT", 3)
print("Sala:", aula_exemplo.sala)
print("Bloco:", aula_exemplo.bloco)
```
