# Análise e Preparação de Dados para o Agendamento de Horários (data.py)

Este notebook tem como objetivo analisar o ficheiro data.py.  
A sua principal função é carregar os dados brutos de um ficheiro de texto (dataset.txt), processá-los utilizando um parser e, por fim, estruturá-los em variáveis Python que serão utilizadas pelo restante do projeto para resolver o problema de agendamento de horários.

## Leitura e parsing do dataset
Esta função lê o ficheiro de texto dataset.txt e converte-o em dicionários Python

```python
from parser import parse_dataset

with open("dataset.txt", "r", encoding="utf-8") as f:
    dataset = f.read()

parsed_data = parse_dataset(dataset)
```

## Listagem de todas as UCs  
Esta função é responsável por agregar todas as unidades curriculares (UCs) existentes no dataset.

```python
def listar_ucs(parsed_data):
    ucs = set()
    for cursos in parsed_data["class_courses"].values():
        ucs.update(cursos)
    for cursos in parsed_data["teacher_courses"].values():
        ucs.update(cursos)
    return sorted(ucs)
```

## Constantes globais  
Definição das listas que serão usadas pelo solver

```python
TURMAS = list(parsed_data["class_courses"].keys())
PROFESSORES = list(parsed_data["teacher_courses"].keys())
BLOCOS = list(range(1, 21))
SALAS = parsed_data["rooms"]
UCS = listar_ucs(parsed_data)
```

## Mapeamento entre Professores, UCs e Turmas  
Finalmente, este script define uma estrutura de dados (professores_por_turma) e uma função auxiliar (achar_turma_por_prof_e_uc). Este mapeamento é crucial porque os dados originais indicam as UCs que um professor leciona, mas não especificam a que turma. Esta estrutura de dados preenche essa lacuna.

```python
professores_por_turma = {
    "LESI": {"PDM": "João", "ISI": "Pedro", "IA": "Pedro", "SETR": "António", "PA": "Manuel"},
    "LEEC": {"PS": "João", "IM": "João", "R": "Pedro", "IE": "Ant   ónio", "RCSD": "Manuel"},
    "LEIM": {"GSI": "João", "IA": "António", "RCE": "António", "AST": "Manuel", "ISC": "Manuel", "GUS": "Manuel"},
    "LDJG": {"IAAJ": "Ana", "PVR": "Isabel", "TCGEV": "Isabel", "PAG": "Ana"}
}

def achar_turma_por_prof_e_uc(professor, uc):
    for turma, ucs in professores_por_turma.items():
        if uc in ucs and ucs[uc] == professor:
            return turma
    return None
```
