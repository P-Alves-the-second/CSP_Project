# Parser do Dataset do Projeto CSP

Este notebook contém o código responsável por **ler e interpretar o ficheiro `dataset.txt`**,  
convertendo-o num formato estruturado (dicionário) que pode ser utilizado nas etapas seguintes do projeto.

## Estrutura do dataset

O ficheiro `dataset.txt` contém várias secções, cada uma iniciada com uma “hashtag” (`#`):

- `#cc` — cursos atribuídos a turmas  
- `#olw` — cursos com apenas uma aula por semana  
- `#dsd` — cursos atribuídos a docentes  
- `#tr` — restrições de horários dos docentes  
- `#rr` — restrições de salas para certas disciplinas  
- `#oc` — aulas online  
- `#rooms` — lista de salas disponíveis  

Linhas começadas com `#head` ou vazias são ignoradas.


```python
from pprint import pprint

def parse_dataset(dataset_text: str):
    """
    Função que interpreta o texto do dataset e devolve um dicionário estruturado.

    Parâmetros:
        dataset_text (str): conteúdo do ficheiro dataset.txt.

    Retorna:
        dict: estrutura com as secções interpretadas.
    """
    sections = {}
    current_section = None
    lines = dataset_text.strip().splitlines()

    for line in lines:
        line = line.strip()
        if not line or line.startswith("#head"):
            continue
        
        if line.startswith("#"):
            current_section = line[1:].split("—")[0].strip()
            sections[current_section] = []
            continue

        if current_section:
            sections[current_section].append(line)

    parsed = {}

    if "cc" in sections:
        class_courses = {}
        for l in sections["cc"]:
            parts = l.split()
            class_courses[parts[0]] = parts[1:]
        parsed["class_courses"] = class_courses

    if "olw" in sections:
        parsed["one_lesson_week"] = [l.strip() for l in sections["olw"]]

    if "dsd" in sections:
        teacher_courses = {}
        for l in sections["dsd"]:
            parts = l.split()
            teacher_courses[parts[0]] = parts[1:]
        parsed["teacher_courses"] = teacher_courses

    if "tr" in sections:
        teacher_restrictions = {}
        for l in sections["tr"]:
            parts = l.split()
            teacher_restrictions[parts[0]] = list(map(int, parts[1:]))
        parsed["teacher_restrictions"] = teacher_restrictions

    if "rr" in sections:
        room_restrictions = {}
        for l in sections["rr"]:
            parts = l.split()
            room_restrictions[parts[0]] = parts[1]
        parsed["room_restrictions"] = room_restrictions

    if "oc" in sections:
        online_classes = {}
        for l in sections["oc"]:
            parts = l.split()
            online_classes[parts[0]] = int(parts[1])
        parsed["online_classes"] = online_classes

    if "rooms" in sections:
        rooms = []
        for l in sections["rooms"]:
            rooms.append(l)
        parsed["rooms"] = rooms

    return parsed
```

## Leitura do ficheiro `dataset.txt`

Nesta célula, o ficheiro é aberto e o seu conteúdo é lido como texto.  
Em seguida, o texto é passado para a função `parse_dataset()` para ser processado.


```python
with open("dataset.txt", "r", encoding="utf-8") as f:
    dataset_text = f.read()

resultado = parse_dataset(dataset_text)
pprint(resultado)
```

## Verificação rápida

Aqui podemos confirmar se as principais secções foram identificadas corretamente.


```python
assert "class_courses" in resultado, "Secção #cc não encontrada!"
assert "teacher_courses" in resultado, "Secção #dsd não encontrada!"
assert "rooms" in resultado, "Secção #rooms não encontrada!"
print("Parser executado com sucesso ✅")

print("Resultado final do parser:\n")
pprint(resultado)

```
